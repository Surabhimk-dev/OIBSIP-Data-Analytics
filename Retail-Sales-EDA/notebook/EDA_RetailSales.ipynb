```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

```python
df.info()
```

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

category_sales
```

```python
plt.figure(figsize=(8,5))

category_sales.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.show()
```

### Business Insight

Technology generates the highest sales among the three categories, indicating strong customer demand for technology products. Furniture contributes substantial sales as well, while Office Supplies has comparatively lower sales.

The business can focus on maintaining strong inventory for Technology products while identifying ways to improve sales in the lower-performing categories.

```python
region_sales = df.groupby("Region")["Sales"].sum()

region_sales.plot(kind="bar", figsize=(8,5))

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.show()
```

```python
top_products = (
    df.groupby("Sub-Category")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_products
```

```python
plt.figure(figsize=(10,6))

top_products.plot(kind="bar")

plt.title("Top 10 Sub-Categories by Sales")
plt.xlabel("Sub-Category")
plt.ylabel("Sales")

plt.show()
```

```python
df["Month"] = df["Order Date"].dt.month_name()
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()

monthly_sales
```

```python
plt.figure(figsize=(12,6))

monthly_sales.plot(marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
# Quarterly Sales Trend

df["Quarter"] = df["Order Date"].dt.to_period("Q")

quarterly_sales = df.groupby("Quarter")["Sales"].sum()

plt.figure(figsize=(10, 5))

quarterly_sales.plot(kind="bar")

plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Sales")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Business Insight

Sales vary across the four quarters, with stronger sales in the later part of the year. This suggests increased customer purchasing activity toward the end of the year, which may be influenced by seasonal demand and holiday shopping.

The business can use this pattern to plan inventory, marketing campaigns, and staffing ahead of high-demand periods.

```python
numeric_df = df.select_dtypes(include=["number"])

plt.figure(figsize=(10,8))

sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8,5))

profit_category = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)

sns.barplot(x=profit_category.index, y=profit_category.values)

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.show()
```

Business Insight:

Technology generates both the highest sales and the highest profit.
Furniture has relatively high sales but much lower profit, indicating lower profit margins or higher discounting.

```python
top10 = (
    df.groupby("Sub-Category")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(12,6))

sns.barplot(x=top10.values, y=top10.index)

plt.title("Top 10 Sub-Categories by Sales")
plt.xlabel("Sales")
plt.ylabel("Sub-Category")

plt.show()
```

Business Insight:

Phones are the best-selling sub-category.
Chairs and Tables generate strong sales, but you should compare them with profit because high sales don't always mean high profitability.

```python
# Top 10 Best-Selling Products

top_products = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))

top_products.sort_values().plot(kind="barh")

plt.title("Top 10 Best-Selling Products")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")

plt.tight_layout()
plt.show()
```

### Business Insight

The top 10 products contribute a significant portion of the overall sales. These products can be prioritized for inventory planning and promotional activities.

The business can also analyze the characteristics of these high-performing products to identify opportunities for cross-selling and recommending similar products to customers.

```python
plt.figure(figsize=(10,5))

sns.histplot(df["Sales"], bins=40)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df["Profit"], bins=40)

plt.title("Profit Distribution")
plt.xlabel("Profit")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,6))

numeric_df = df.select_dtypes(include="number")

sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

Business Insight:

Excessive discounting negatively affects profitability.
Increasing sales generally increases profit, but discount strategies should be optimized.

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

monthly_sales = (
    df.groupby(df["Order Date"].dt.month)["Sales"]
      .sum()
)

plt.figure(figsize=(12,6))

monthly_sales.plot(marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(range(1,13))

plt.show()
```

```python
plt.figure(figsize=(8,5))

segment_sales = df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)

sns.barplot(x=segment_sales.index, y=segment_sales.values)

plt.title("Sales by Customer Segment")
plt.xlabel("Segment")
plt.ylabel("Sales")

plt.show()
```

```python
top_customers = (
    df.groupby("Customer Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,6))

sns.barplot(x=top_customers.values, y=top_customers.index)

plt.title("Top 10 Customers by Sales")
plt.xlabel("Sales")
plt.ylabel("Customer")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(data=df, x="Sales", y="Profit")

plt.title("Sales vs Profit")

plt.show()
```

# Conclusion

The exploratory data analysis provided useful insights into sales, profit, product performance, categories, regions, and sales trends over time.

The analysis shows that sales performance varies across categories, regions, products, and periods. These insights can help the business improve sales performance and make better inventory and marketing decisions.

## Actionable Business Recommendations

1. **Focus on high-performing products and categories:**  
   Maintain sufficient inventory for the best-selling products and strongest categories to avoid stock shortages during periods of high demand.

2. **Improve sales in weaker-performing categories and regions:**  
   Identify the reasons for lower sales and use targeted promotions, discounts, and marketing campaigns to increase customer demand in those areas.

3. **Plan inventory and marketing around seasonal trends:**  
   Use the monthly and quarterly sales trends to prepare inventory and launch marketing campaigns before periods of higher customer demand.

4. **Promote profitable products:**  
   Give greater attention to products and categories that generate strong profit, rather than focusing only on sales volume.

5. **Use customer and regional insights for targeted marketing:**  
   Customize promotional strategies based on customer segments and regional sales performance to improve engagement and revenue.

Overall, the analysis can support better decisions related to inventory management, marketing, product strategy, and sales planning.
